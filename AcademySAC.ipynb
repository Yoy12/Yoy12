# Sports Finance Data Workshop 😎😎😎😎😎
Let's simulate a real-world sports data scenario. You have been given incomplete, messy financial data. Your goal is to clean it, engineer new features, and match the data to the correct player profile.

```python
import pandas as pd
import numpy as np

# Generate synthetic messy data
np.random.seed(42)
data = {
    'Player_ID': ['Mbappe', 'Player_B', 'Player_C', 'Player_D', 'Player_E'],
    'Goals': [33, 12, 5, 45, 2],
    'Assists': [6, 15, 2, 10, 1],
    'Trophy_Bonus': ['Not Met', 'Met', 'Not Met', 'Met', 'Not Met'],
    'International_Bonus': ['Met', 'Not Met', 'Met', 'Met', 'Not Met'],
    'Games_Played': [40, 38, 400, 42, 15], # Note the 400 outlier
    'Jerseys_Sold_Millions': [300, 45, np.nan, 500, 2], # Note the NaN
    'Salary_Amount_Euros': ['€150M', '€40M', '€10M', '€200M', '€5M']
}
df = pd.DataFrame(data)
display(df)
```

## Phase 1: The Messy Locker Room
### 1.1 Handling Missing Values

```python
# TODO: Fill missing values (NaNs) in the 'Jerseys_Sold_Millions' column with the median of that column.

df['Jerseys_Sold_Millions'] = df['Jerseys_Sold_Millions'].fillna(df['Jerseys_Sold_Millions'].median())
display(df)
```

### 1.2 Fixing Data Types

```python
# TODO: The 'Salary_Amount_Euros' is a string with '€' and 'M'.
# 1. Remove the '€' and 'M' characters.
# 2. Convert the column to a float.
# 3. Multiply by 1,000,000 to get the actual numeric salary.

df['Salary_Amount_Euros'] = df['Salary_Amount_Euros'].replace({'€':'', 'M':''}, regex=True).astype(float) * 1000000
display(df)
```

### 1.3 Outlier Detection

```python
# TODO: There is a typo in 'Games_Played'. Filter the DataFrame to only include rows where 'Games_Played' is a realistic number (e.g., <= 70).

df = df[df['Games_Played'] <= 70]
display(df)
```

## Phase 2: The Moneyball Phase (Feature Engineering)
### 2.1 Composite Metrics

```python
# TODO: Create a new column called 'Total_Goal_Contributions' which is the sum of 'Goals' and 'Assists'.
df['Total_Goal_Contributions'] = df['Goals'] + df['Assists']
display(df)
```

### 2.2 Efficiency Metrics

```python
# TODO: Create a 'Salary_Per_Goal' column by dividing 'Salary_Amount_Euros' by 'Goals'.
# Tip: Be careful not to divide by zero if a player has 0 goals!
df['Salary_Per_Goal'] = df['Salary_Amount_Euros'] / df['Goals'].replace(0, 1)
display(df)
```

### 2.3 Boolean Encoding

```python
# TODO: Convert the 'Trophy_Bonus' column from 'Met'/'Not Met' strings into integers (1s and 0s) in a new column called 'Trophy_Bonus_Met'.
df['Trophy_Bonus_Met'] = (df['Trophy_Bonus'] == 'Met').astype(int)
display(df)
```

## Phase 3: The Contract Reveal (Match the Player)
Can we find the player with 33 goals, €150M salary, and 300M jerseys sold?

```python
# TODO: Filter the DataFrame to find the player who matches our mystery profile:
# - Goals == 33
# - Salary_Amount_Euros == 150000000
mystery_player = df[(df['Goals'] == 33) & (df['Salary_Amount_Euros'] == 150000000)]
display(mystery_player)
```
